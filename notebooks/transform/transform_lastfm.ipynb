```python
import json
import pandas as pd
```

```python
df_lastfm = pd.read_json("../../data/raw/lastfm_scrobbles.json")
df_lastfm.head(10)
```

```python
df_lastfm.shape
```

```python
artist_normalizado = pd.json_normalize(df_lastfm['artist'])
album_normalizado = pd.json_normalize(df_lastfm['album'])
data_normalizado = pd.json_normalize(df_lastfm['date'])
df_lastfm['artist_name'] = artist_normalizado['#text']
df_lastfm['album_name'] = album_normalizado['#text']
df_lastfm['scrobble_date'] = data_normalizado['uts'] 
df_lastfm['music_name'] = df_lastfm['name']
# trocando a extração de text para uts -> timestamp unix que contabiliza as horas desde 00:00 1 de janeiro de 1970
df_lastfm['artist_mbid'] = df_lastfm['artist'].str['mbid'] #extraindo ID
df_lastfm.head()
```

```python
df_lastfm.drop(columns=['name','image','artist','streamable','mbid','date','album','url'], inplace=True)
```

```python
df_lastfm.head(10)
```

```python
df_lastfm.columns
```

```python
print(df_lastfm['scrobble_date'].dtype)
```

```python
df_lastfm['scrobble_date'] = df_lastfm['scrobble_date'].astype(int) #unit só lê int, e uts estava como str
```

```python
#ajustes a serem feitos:
# Converter scrobble_date de texto pra um tipo datetime de verdade
# Ajustar o fuso horário de UTC pra Brasília (UTC-3)

df_lastfm['scrobble_date'] = pd.to_datetime(df_lastfm['scrobble_date'], unit='s', utc=True)
#unit='s' indica que o número indica segundos
#utc=True diz que essa data, assim que convertida, esta marcada como UTC -> padrão global para registrar datas e horas sem considerar fuso-horário
# primeiro passar para padrão (utc), depois passar para fuso de brasília
```

```python
df_lastfm['scrobble_date'] = df_lastfm['scrobble_date'].dt.tz_convert('America/Sao_Paulo')
df_lastfm.head(10)
```

```python
df_lastfm['date'] = df_lastfm['scrobble_date'].dt.date
df_lastfm['time'] = df_lastfm['scrobble_date'].dt.time
df_lastfm['week_day'] = df_lastfm['scrobble_date'].dt.day_name()
df_lastfm['week_day_id'] = df_lastfm['scrobble_date'].dt.dayofweek
df_lastfm['group_hour'] = df_lastfm['scrobble_date'].dt.hour
# melhor para fazer agrupamento dos tempos, pois cada tempo é diferente -> mais difícil para análise

df_lastfm.head(10)

#mantém scroble_date para comparações mais fáceis de serem feitas que os dados separados
```

```python
df_lastfm.to_csv('../../data/processed/lastfm_transformed.csv', index=False)
```
