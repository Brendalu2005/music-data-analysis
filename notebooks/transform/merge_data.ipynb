```python
import pandas as pd
```

```python
df_lastfm = pd.read_csv('../../data/processed/lastfm_transformed.csv')
df_lastfm.head(10)
```

```python
df_mb= pd.read_csv('../../data/processed/musicBrainz_transformed.csv')
```

```python
print(df_lastfm.columns.tolist())
```

```python
print(df_mb.columns.tolist())
```

```python
# Padroniza os nomes: remove espaços nas pontas e converte para minúsculas
df_lastfm['nome_chave'] = df_lastfm['artist_name'].str.strip().str.lower()
df_mb['nome_chave'] = df_mb['query_name'].str.strip().str.lower()
# padroniza nomes de artistas ex: willow, que no musicBrainz aparecia como WILLOW

# Mantém todas as linhas da tabela Last.fm
df_final = df_lastfm.merge(
    df_mb,
    on='nome_chave',
    how='left'
)

# Remove a coluna auxiliar do resultado
df_final = df_final.drop(columns=['nome_chave'])
# query_name é a "ponte" garantida entre as duas tabelas, porque ela nasceu do mesmo valor original.
```
